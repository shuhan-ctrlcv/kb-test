# Yield Analysis

PedalWorks' Quality stage fails the very first unit of every run on purpose: the first-article inspection always fails, and that unit is reworked at Assembly before Quality re-inspects it. This notebook estimates the resulting first-pass yield for a City Cruiser production run and compares it against the Trail Blazer, which goes through the same Quality gate after Assembly consumes a Frame, Wheelset, and Suspension Fork instead of a Drivetrain and Brake Set.

```python
# Toy run: units inspected, first-article always fails and is reworked.
units_per_run = 50
first_article_failures = 1  # PedalWorks always fails the first unit off a run
first_pass_yield = (units_per_run - first_article_failures) / units_per_run
print(f"First-pass yield: {first_pass_yield:.1%}")
```
